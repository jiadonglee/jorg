# Korg vs Jorg comparison

Run this notebook from the Korg.jl repo root. It computes spectra with Jorg and Korg, then compares outputs for a given set of stellar parameters.

```python
# Parameters
Teff = 5779
logg = 4.5
m_H = 0.0
wl_min = 5000.0
wl_max = 5020.0
# wl_min = 3800.0
# wl_max = 3820.0
# wl_min = 8900.0
# wl_max = 8920.0
hydrogen_lines = False

tag = f"Teff{Teff:.0f}_g{logg:.2f}_mH{m_H:.2f}"
tag = tag.replace('.', 'p').replace('-', 'm')
tag
```

```python
import subprocess
import sys
from pathlib import Path

import numpy as np
import gzip
import plotly.graph_objects as go

# Find repo root by walking up from notebook location
repo_root = Path.cwd()
while repo_root != repo_root.parent:
    if (repo_root / 'jorg' / 'src').exists() and (repo_root / 'data').exists():
        break
    repo_root = repo_root.parent
else:
    raise RuntimeError('Could not find Korg.jl repo root. Expected jorg/src and data directories.')

sys.path.insert(0, str(repo_root / 'jorg' / 'src'))
sys.path.insert(0, str(repo_root / 'kurucz'))

from jorg.synthesis import synthesize, create_korg_compatible_abundance_array
from jorg.lines.linelist import read_linelist
from jorg.atmosphere import interpolate_marcs

try:
    import kurucz
    kurucz_available = True
except Exception as exc:
    kurucz_available = False
    print(f'Kurucz import failed: {exc}')

linelist_path = repo_root / 'data' / 'linelists' / 'vald_extract_stellar_solar_threshold001.vald'
korg_script = repo_root / 'jorg' / 'korg_script' / 'korg_compare_params.jl'
out_dir = repo_root / 'comparison_outputs'
out_dir.mkdir(exist_ok=True)

linelist_path
```

```python
# Jorg synthesis
linelist = read_linelist(str(linelist_path), format='vald')
atm = interpolate_marcs(Teff, logg, m_H)
A_X = create_korg_compatible_abundance_array(m_H)

result_lines = synthesize(
    atm,
    linelist,
    A_X,
    wavelengths=(wl_min, wl_max),
    hydrogen_lines=hydrogen_lines,
    verbose=False,
    logg=logg,
)
result_cntm = synthesize(
    atm,
    [],
    A_X,
    wavelengths=(wl_min, wl_max),
    hydrogen_lines=hydrogen_lines,
    verbose=False,
    logg=logg,
)

wl = np.asarray(result_lines.wavelengths)
flux = np.asarray(result_lines.flux)
cntm = np.asarray(result_lines.cntm)
rect = flux / cntm

wl_c = np.asarray(result_cntm.wavelengths)
flux_c = np.asarray(result_cntm.flux)
cntm_c = np.asarray(result_cntm.cntm)
rect_c = flux_c / cntm_c

np.savetxt(out_dir / f'jorg_{tag}_spectrum_with_lines.txt', np.column_stack([wl, flux, cntm, rect]))
np.savetxt(out_dir / f'jorg_{tag}_spectrum_continuum.txt', np.column_stack([wl_c, flux_c, cntm_c, rect_c]))
np.savetxt(out_dir / f'jorg_{tag}_alpha_with_lines.txt', np.asarray(result_lines.alpha))
np.savetxt(out_dir / f'jorg_{tag}_alpha_continuum.txt', np.asarray(result_cntm.alpha))

out_dir
```

```python
import time

from jorg.opacity.layer_processor import LayerProcessor
from jorg.synthesis import _calculate_line_opacity_multilayer, _calculate_radiative_transfer
from jorg.statmech import (
    create_default_partition_functions,
    create_default_ionization_energies,
    create_default_log_equilibrium_constants,
)
from jorg.alpha5_reference import calculate_alpha5_reference

# Match synthesize_korg_compatible wavelength grid
spacing = 0.01
n_points = int((wl_max - wl_min) / spacing) + 1
wl_array = np.linspace(wl_min, wl_max, n_points)

# Convert atmosphere to dict if needed (matches synthesize_korg_compatible)
if hasattr(atm, 'layers'):
    atm_dict = {
        'temperature': np.array([layer.temp for layer in atm.layers]),
        'electron_density': np.array([layer.electron_number_density for layer in atm.layers]),
        'number_density': np.array([layer.number_density for layer in atm.layers]),
        'tau_5000': np.array([layer.tau_5000 for layer in atm.layers]),
        'height': np.array([layer.z for layer in atm.layers]),
    }
    atm_dict['pressure'] = atm_dict['number_density'] * 1.380649e-16 * atm_dict['temperature']
else:
    atm_dict = atm

abs_abundances = 10 ** (A_X - 12)
abs_abundances = abs_abundances / np.sum(abs_abundances)
abs_abundances_dict = {Z: abs_abundances[Z - 1] for Z in range(1, 93)}

ionization_energies = create_default_ionization_energies()
partition_funcs = create_default_partition_functions()
log_equilibrium_constants = create_default_log_equilibrium_constants()

layer_processor = LayerProcessor(
    ionization_energies=ionization_energies,
    partition_funcs=partition_funcs,
    log_equilibrium_constants=log_equilibrium_constants,
    line_cutoff_threshold=3e-4,
    verbose=False,
)

# Stage 1: chemical equilibrium + continuum
stage1_start = time.perf_counter()
alpha_continuum, all_number_densities, all_electron_densities = layer_processor.process_all_layers(
    atm=atm_dict,
    abs_abundances=abs_abundances_dict,
    wl_array=wl_array,
    linelist=None,
    line_buffer=10.0,
    hydrogen_lines=False,
    vmic=1.0,
    use_chemical_equilibrium_from=None,
    log_g=logg,
)
stage1_end = time.perf_counter()

# Stage 2: line opacity
stage2_start = time.perf_counter()
line_opacity = _calculate_line_opacity_multilayer(
    wl_array=wl_array,
    temps=np.array(atm_dict['temperature']),
    electron_densities=all_electron_densities,
    number_densities=all_number_densities,
    partition_funcs=partition_funcs,
    linelist=linelist,
    line_buffer=10.0,
    microturbulence_kms=1.0,
    continuum_opacity=alpha_continuum,
    cutoff_threshold=3e-4,
    verbose=False,
)
stage2_end = time.perf_counter()

alpha_matrix = alpha_continuum + line_opacity

# Alpha5 reference
alpha5_start = time.perf_counter()
_ = calculate_alpha5_reference(
    atm,
    A_X,
    linelist=linelist,
    number_densities=all_number_densities,
    electron_densities=all_electron_densities,
    partition_funcs=partition_funcs,
    microturbulence_kms=1.0,
    line_cutoff_threshold=3e-4,
    verbose=False,
)
alpha5_end = time.perf_counter()

# Radiative transfer
rt_start = time.perf_counter()
_ = _calculate_radiative_transfer(
    alpha_matrix,
    atm_dict,
    wl_array,
    result_lines.mu_grid,
    'linear_flux_only',
    True,
    A_X,
    layer_processor,
    linelist,
    10.0,
    hydrogen_lines,
    1.0,
    abs_abundances,
    None,
    logg,
    False,
    rt_method='korg_default',
    verbose=False,
    alpha_continuum=alpha_continuum,
    line_cutoff_threshold=3e-4,
)
rt_end = time.perf_counter()

print(f"Timing Stage1 (chem eq + continuum): {stage1_end - stage1_start:.2f}s")
print(f"Timing Stage2 (line opacity): {stage2_end - stage2_start:.2f}s")
print(f"Timing Alpha5 reference: {alpha5_end - alpha5_start:.2f}s")
print(f"Timing RT: {rt_end - rt_start:.2f}s")
```

```python
korg_script = repo_root / 'jorg' / 'korg_script' / 'korg_compare_params.jl'

# Korg synthesis via Julia
cmd = [
    'julia',
    '--project=.',
    str(korg_script),
    str(Teff),
    str(logg),
    str(m_H),
    str(wl_min),
    str(wl_max),
    tag,
    str(out_dir),
]
subprocess.run(cmd, cwd=str(repo_root), check=True)

out_dir
```

```python
# Load outputs
k_lines = np.loadtxt(out_dir / f'korg_{tag}_spectrum_with_lines.txt')
j_lines = np.loadtxt(out_dir / f'jorg_{tag}_spectrum_with_lines.txt')
k_cntm = np.loadtxt(out_dir / f'korg_{tag}_spectrum_continuum.txt')
j_cntm = np.loadtxt(out_dir / f'jorg_{tag}_spectrum_continuum.txt')
k_alpha_lines = np.loadtxt(out_dir / f'korg_{tag}_alpha_with_lines.txt')
k_alpha_cntm = np.loadtxt(out_dir / f'korg_{tag}_alpha_continuum.txt')
j_alpha_lines = np.loadtxt(out_dir / f'jorg_{tag}_alpha_with_lines.txt')
j_alpha_cntm = np.loadtxt(out_dir / f'jorg_{tag}_alpha_continuum.txt')
k_lines.shape, j_lines.shape, k_cntm.shape, j_cntm.shape
```

```python
import importlib, kurucz
import struct
print(kurucz.__file__)
importlib.reload(kurucz)


# Kurucz synthesis (fast mode) and load spectrum
kurucz_rect = None
kurucz_wl = None
kurucz_flux = None


def _kurucz_has_valid_fort93(path):
    if not path.exists():
        return False
    try:
        raw = path.read_bytes()
        if len(raw) < 12:
            return False
        rec_len = struct.unpack('<i', raw[:4])[0]
        if rec_len + 8 > len(raw):
            return False
        rec = raw[4:4 + rec_len]
        ofs = 0
        ofs += 5 * 4  # NLINES, LENGTH, IFVAC, IFNLTE, N19
        ofs += 4      # TURBV (real*4)
        ofs += 4 * 7 * 99  # DECKJ
        ofs += 4      # IFPRED
        if ofs + 5 * 8 > len(rec):
            return False
        wlbeg, wlend, resolu, ratio, ratiolg = struct.unpack_from('<5d', rec, ofs)
        return resolu > 0 and wlend > wlbeg
    except Exception:
        return False


def _kurucz_target_wavelengths():
    if 'k_lines' in globals():
        wl = k_lines[:, 0]
        return float(np.nanmin(wl)), float(np.nanmax(wl))
    if 'j_lines' in globals():
        wl = j_lines[:, 0]
        return float(np.nanmin(wl)), float(np.nanmax(wl))
    return float(wl_min), float(wl_max)


def _is_lfs_pointer(path):
    try:
        if not path.exists():
            return False
        with open(path, 'r', encoding='ascii', errors='ignore') as handle:
            first_line = handle.readline()
        return first_line.startswith('version https://git-lfs.github.com/spec/v1')
    except OSError:
        return False


if kurucz_available:
    wl_min_aa, wl_max_aa = _kurucz_target_wavelengths()
    kurucz_wl_min_nm = wl_min_aa / 10.0
    kurucz_wl_max_nm = wl_max_aa / 10.0
    kurucz_linelist_dir = f'linelists_{wl_min_aa:.1f}_{wl_max_aa:.1f}'
    kurucz_linelist_dir = kurucz_linelist_dir.replace('.', 'p').replace('-', 'm')
    kurucz_linelist_path = repo_root / 'kurucz' / 'synthe' / kurucz_linelist_dir
    fort93_path = kurucz_linelist_path / 'tfort.93'
    gfall_path = repo_root / 'kurucz' / 'lines' / 'gfallvac.latest'

    if _is_lfs_pointer(gfall_path):
        print(f'Kurucz line list is a Git LFS pointer: {gfall_path}')
        print('Run: git lfs pull')
        spec_file = None
    else:
        try:
            needs_setup = (not kurucz_linelist_path.exists()) or (not _kurucz_has_valid_fort93(fort93_path))
            if fort93_path.exists() and gfall_path.exists():
                needs_setup = needs_setup or (gfall_path.stat().st_mtime > fort93_path.stat().st_mtime)
            if needs_setup:
                print(f'Kurucz line lists missing or invalid at {kurucz_linelist_path}, running setup...')
                kurucz.setup(wl_start=kurucz_wl_min_nm, wl_end=kurucz_wl_max_nm, output_dir=kurucz_linelist_dir)

            kurucz_tag = f'kfast_{Teff:.0f}_{logg:.2f}_{m_H:.2f}'
            kurucz_tag = kurucz_tag.replace('.', 'p').replace('-', 'm')

            kurucz_debug = True
            kurucz_keep_tmp = True

            spec_file = kurucz.synthe_fast(
                kurucz_tag,
                teff=Teff,
                logg=logg,
                mh=m_H,
                am=0.0,
                vturb=vmic if 'vmic' in globals() else 2.0,
                line_list_dir=kurucz_linelist_dir,
                debug=kurucz_debug,
                keep_tmp=kurucz_keep_tmp,
            )
        except kurucz.KuruczError as exc:
            print(f'Kurucz setup/synthesis failed: {exc}')
            spec_file = None
        except FileNotFoundError as exc:
            print('Kurucz synthe_fast failed to create the temp model file.')
            print('Try a shorter kurucz_tag or check kurucz/bin_* executables and permissions.')
            print(f'Error: {exc}')
            spec_file = None

    if spec_file is None:
        print('Kurucz synthesis failed; skipping Kurucz comparison.')
    else:
        with gzip.open(spec_file, 'rt') as handle:
            kurucz_data = np.loadtxt(handle)
        kurucz_wl = kurucz_data[:, 0] * 10.0  # nm -> Angstrom
        kurucz_flux = kurucz_data[:, 1]
        kurucz_cntm = kurucz_data[:, 2]
        kurucz_rect = kurucz_flux / kurucz_cntm
        print(f'Kurucz spectrum loaded: {kurucz_wl.min():.2f}-{kurucz_wl.max():.2f} A')
```

### Rectification note
The continuum-only run is flat at 1.0 for both codes, so rectification is being applied. In 3800-4000 A, Jorg's continuum sits higher than Korg's while its flux increase is smaller, so the flux/continuum ratio stays below 1.0 across the window. That points to stronger line blanketing or continuum opacity differences, not a missing normalization step.


```python
# Line density in the window (VALD format; column 2 is wavelength in A)
wl_min = 3800.0
wl_max = 4000.0
line_buffer = 10.0
parsed_lines = 0
lines_in_window = 0
lines_in_window_buffer = 0
with open(linelist_path, 'r') as handle:
    for raw in handle:
        raw = raw.strip()
        if not raw.startswith("'"):
            continue
        parts = [p.strip() for p in raw.split(',')]
        if len(parts) < 2:
            continue
        try:
            wl = float(parts[1])
        except ValueError:
            continue
        parsed_lines += 1
        if wl_min <= wl <= wl_max:
            lines_in_window += 1
        if (wl_min - line_buffer) <= wl <= (wl_max + line_buffer):
            lines_in_window_buffer += 1
print(f"Parsed VALD lines: {parsed_lines}")
print(f"Lines in {wl_min:.0f}-{wl_max:.0f} A: {lines_in_window}")
print(
    f"Lines in {wl_min-line_buffer:.0f}-{wl_max+line_buffer:.0f} A "
    f"(with buffer): {lines_in_window_buffer}"
)
```

```python
# Rough scale of line-opacity work
n_layers = j_alpha_lines.shape[0]
n_wl = j_alpha_lines.shape[1]
print(f"Layers: {n_layers}, wavelengths: {n_wl}")
print(f"Line-layer pairs: ~{lines_in_window_buffer * n_layers:,}")
```

### Work estimate note
Each contributing line is evaluated across a window of wavelength points for every layer in Python. The line-layer count above is a lower bound; the real profile evaluations scale with the per-line window size, so runtime is dominated by line processing and only weakly by the overall wavelength grid length.


### Performance note
The line opacity step in `jorg/src/jorg/opacity/korg_line_processor.py` uses Python-level nested loops (lines x layers x wavelength points) to evaluate Voigt profiles. In 3800-4000 A there are thousands of VALD lines, so runtime is dominated by per-line processing and broadening, not the wavelength grid spacing. Korg.jl runs the same algorithm in compiled Julia, which explains the ~4 min vs ~43.5 s gap. Fixed-cost steps like chemical equilibrium and linelist parsing also add grid-independent overhead.


```python
# Diagnostics
n_layers = min(k_alpha_lines.shape[0], j_alpha_lines.shape[0])
n_wl = min(k_alpha_lines.shape[1], j_alpha_lines.shape[1])

k_alpha_lines = k_alpha_lines[:n_layers, :n_wl]
k_alpha_cntm = k_alpha_cntm[:n_layers, :n_wl]
j_alpha_lines = j_alpha_lines[:n_layers, :n_wl]
j_alpha_cntm = j_alpha_cntm[:n_layers, :n_wl]

k_line_only = k_alpha_lines - k_alpha_cntm
j_line_only = j_alpha_lines - j_alpha_cntm

k_line_mean = np.mean(np.abs(k_line_only), axis=1)
j_line_mean = np.mean(np.abs(j_line_only), axis=1)

line_rel_diff = np.mean(np.abs(j_line_mean - k_line_mean) / np.maximum(k_line_mean, 1e-40))

flux_scale = np.mean(j_lines[:, 1]) / np.mean(k_lines[:, 1])
rect_mean_abs = np.mean(np.abs(j_lines[:, 3] - k_lines[:, 3]))

print(f'mean |line alpha| relative diff: {line_rel_diff:.3e}')
print(f'flux scale factor (Jorg/Korg): {flux_scale:.6f}')
print(f'rectified mean abs diff: {rect_mean_abs:.6f}')
```

```python
# Plots
layer_idx = np.arange(n_layers)
eps = 1e-40

fig = go.Figure()
fig.add_trace(go.Scatter(x=layer_idx, y=np.log10(k_line_mean + eps), mode='lines', name='Korg |mean line alpha|'))
fig.add_trace(go.Scatter(x=layer_idx, y=np.log10(j_line_mean + eps), mode='lines', name='Jorg |mean line alpha|'))
fig.update_layout(
    template='plotly_white',
    width=850,
    height=450,
    xaxis_title='Layer index',
    yaxis_title='log10(|mean line alpha|) [cm^-1]',
    title='Line-only opacity per layer'
)
fig.show()

fig = go.Figure()
fig.add_trace(go.Scatter(x=k_lines[:, 0], y=k_lines[:, 1], mode='lines', name='Korg flux'))
fig.add_trace(go.Scatter(x=j_lines[:, 0], y=j_lines[:, 1], mode='lines', name='Jorg flux'))
fig.add_trace(go.Scatter(x=j_lines[:, 0], y=j_lines[:, 1] / flux_scale, mode='lines', name=f'Jorg flux / {flux_scale:.3f}', line=dict(dash='dash')))
if kurucz_flux is not None:
    kurucz_flux_interp = np.interp(k_lines[:, 0], kurucz_wl, kurucz_flux)
    fig.add_trace(go.Scatter(x=k_lines[:, 0], y=kurucz_flux_interp, mode='lines', name='Kurucz flux'))
fig.update_layout(
    template='plotly_white',
    width=850,
    height=450,
    xaxis_title='Wavelength [A]',
    yaxis_title='Flux [erg/s/cm^2/cm]',
    title='Flux normalization (with lines)'
)
fig.show()

fig = go.Figure()
fig.add_trace(go.Scatter(x=k_lines[:, 0], y=k_lines[:, 3], mode='lines', name='Korg rectified'))
fig.add_trace(go.Scatter(x=j_lines[:, 0], y=j_lines[:, 3], mode='lines', name='Jorg rectified', line=dict(dash='dot')))
if kurucz_rect is not None:
    kurucz_rect_interp = np.interp(k_lines[:, 0], kurucz_wl, kurucz_rect)
    fig.add_trace(go.Scatter(x=k_lines[:, 0], y=kurucz_rect_interp, mode='lines', name='Kurucz rectified'))
fig.update_layout(
    template='plotly_white',
    width=900,
    height=450,
    xaxis_title='Wavelength [A]',
    yaxis_title='Rectified flux',
    title='Rectified spectrum (Korg vs Jorg vs Kurucz)'
)
fig.show()
```
